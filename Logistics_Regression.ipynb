2. What are the key assumptions of Simple Linear Regression?

--> The key assumptions of Simple Linear Regression (SLR) are:

a. Linearity: The relationship between the independent (predictor) variable and the dependent (response) variable is assumed to be linear. This means that the change in the dependent variable is proportional to the change in the independent variable, and the data points should roughly align around a straight line.

b. Independence of Errors: The residuals (errors) from the regression model should be independent of each other, meaning that the error for one observation does not depend on the error for another. This is especially important in time series or ordered data.

c. Homoscedasticity: The variance of the residuals (errors) should be constant across all levels of the independent variable. This means the spread of residuals around the regression line should be roughly equal for all values of the predictor.

d. Normality of Residuals: The residuals should be approximately normally distributed. This assumption is crucial for hypothesis testing and constructing confidence intervals, but less critical for prediction accuracy.

e. Measurement Level: Both the dependent and independent variables should be continuous (interval or ratio scale).

6. What is Logistic Regression? How does it differ from Linear Regression?

--> Logistic Regression is a supervised learning algorithm used for classification tasks where the outcome is categorical, often binary (such as yes/no or success/failure). It predicts the probability that a given input belongs to a certain category by applying a logistic (sigmoid) function, which outputs values between 0 and 1.

Linear Regression, on the other hand, is used for predicting continuous numeric outcomes by modeling a linear relationship between the dependent variable and one or more independent variables.

The key difference is that Linear Regression predicts a continuous value and the relationship between variables is represented by a straight line, while Logistic Regression predicts probabilities that translate into categories using an S-shaped curve. Linear Regression uses the method of least squares for estimating parameters, assuming normally distributed errors, whereas Logistic Regression uses maximum likelihood estimation and assumes the dependent variable follows a binomial distribution.

In essence, Linear Regression is suited for predicting quantities, like house prices or temperatures, while Logistic Regression is used for classification problems, like determining if an email is spam or not, or whether a patient has a disease based on given input features.

Thus, Logistic Regression differs in purpose, output type, mathematical formulation, and how results are interpreted compared to Linear Regression.

7. Name and briefly describe three common evaluation metrics for regression models.

--> Three common evaluation metrics for regression models are:

Mean Absolute Error (MAE): This metric calculates the average of the absolute differences between the predicted values and the actual values. It gives a straightforward interpretation of the average error size, showing how far, on average, the predictions are from the true values without considering the direction of the errors.

Mean Squared Error (MSE): MSE is the average of the squared differences between the predicted and actual values. By squaring the errors, it penalizes larger errors more than smaller ones, making it sensitive to outliers. It's useful when large deviations are particularly undesirable.

R-squared (R²): This is a statistical measure that indicates the proportion of the variance in the dependent variable that is predictable from the independent variable(s). It ranges from 0 to 1, where a higher value means the model explains a greater fraction of the variability in the outcome.

8. What is the purpose of the R-squared metric in regression analysis?

--> The purpose of the R-squared metric in regression analysis is to measure how well the regression model explains the variability of the dependent variable around its mean. It quantifies the proportion of the total variance in the dependent variable that is accounted for by the independent variable(s) in the model.

R-squared values range from 0 to 1, where a value of 1 indicates a perfect fit—meaning the model explains all the variation in the outcome—and a value of 0 means the model explains none of the variation, so the mean of the dependent variable is as good a predictor as the regression model. Essentially, a higher R-squared indicates that the model provides a better fit to the observed data by reducing the unexplained variability.

This metric helps assess and compare the performance of regression models, guiding analysts to understand how effectively the predictors explain the target variable's behavior.

9. Write Python code to fit a simple linear regression model using scikit-learn
and print the slope and intercept.
(Include your Python code and output in the code box below.)

```python
from sklearn.linear_model import LinearRegression
import numpy as np

# Sample data: X = independent variable, y = dependent variable
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([3, 4, 2, 5, 6])

# Create a LinearRegression model
model = LinearRegression()

# Fit the model
model.fit(X, y)

# Print the slope (coefficient) and intercept
print("Slope (Coefficient):", model.coef_[0])
print("Intercept:", model.intercept_)
```
